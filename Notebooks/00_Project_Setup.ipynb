```python
PROJECT_ROOT = Path("/content/drive/MyDrive/WildfireProject")

DATA_DIR = PROJECT_ROOT / "data"
MODELS_DIR = PROJECT_ROOT / "models"
RESULTS_DIR = PROJECT_ROOT / "results"
FIGURES_DIR = PROJECT_ROOT / "figures"

print(PROJECT_ROOT)
```

```python
folders = [
    DATA_DIR / "raw",
    DATA_DIR / "processed",
    DATA_DIR / "simulated",

    MODELS_DIR / "baseline",
    MODELS_DIR / "quantized",
    MODELS_DIR / "checkpoints",

    RESULTS_DIR / "metrics",
    RESULTS_DIR / "logs",
    RESULTS_DIR / "predictions",

    FIGURES_DIR / "week1"
]

for folder in folders:
    folder.mkdir(parents=True, exist_ok=True)

print("Project folders created successfully.")
```

```python
for root, dirs, files in os.walk(PROJECT_ROOT):
    level = root.replace(str(PROJECT_ROOT), '').count(os.sep)
    indent = " " * 4 * level
    print(f"{indent}{os.path.basename(root)}/")
```

```python
SEED = 42

random.seed(SEED)
np.random.seed(SEED)

print("Random seed:", SEED)
```

```python
experiment = {
    "project":
        "Wildfire TinyML Self-Sufficiency",

    "week":
        1,

    "experiment":
        "Environmental Data Simulation",

    "seed":
        SEED,

    "sampling_rate_seconds":
        1,

    "version":
        "0.1"
}

experiment
```

```python
import json

config_path = PROJECT_ROOT / "experiment_config.json"

with open(config_path, "w") as f:
    json.dump(experiment, f, indent=4)

print("Configuration saved.")
```
